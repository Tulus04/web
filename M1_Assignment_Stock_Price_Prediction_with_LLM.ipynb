```python
import yfinance as yf
import pandas as pd

# Download historical data for AAPL if not already present in workspace
try:
    data
except NameError:
    data = yf.download("AAPL", start="2020-01-01", end="2023-01-01")

# Calculate moving averages
data['MA_5'] = data['Close'].rolling(window=5).mean()
data['MA_10'] = data['Close'].rolling(window=10).mean()
data['MA_50'] = data['Close'].rolling(window=50).mean()

# Drop rows with NaN values resulting from the rolling calculations
data_cleaned = data.dropna()

# Display the first 5 rows
print(data_cleaned.head())
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Create binary target: 1 if tomorrow's Close is higher than today's Close, 0 otherwise
data_cleaned['Target'] = (data_cleaned['Close'].shift(-1) > data_cleaned['Close']).astype(int)

# Drop the last row as it will have NaN for the Target
model_data = data_cleaned.dropna()

# Define features and target
features = ['Open', 'High', 'Low', 'Close', 'Volume', 'MA_5', 'MA_10', 'MA_50']
X = model_data[features]
y = model_data['Target']

# Perform a 70/30 train/test split without shuffling to preserve time order
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, shuffle=False)

# Train Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Predict and evaluate
y_pred = rf_model.predict(X_test)

# Print metrics
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Feature Importances
importances = rf_model.feature_importances_
for feature, importance in zip(features, importances):
    print(f"{feature}: {importance:.4f}")
```

```python
# Download out-of-sample data for AAPL
new_data = yf.download("AAPL", start="2023-01-01", end="2024-01-01")

# Calculate the same moving average features
new_data['MA_5'] = new_data['Close'].rolling(window=5).mean()
new_data['MA_10'] = new_data['Close'].rolling(window=10).mean()
new_data['MA_50'] = new_data['Close'].rolling(window=50).mean()

# Drop NaN values resulting from rolling calculations
new_data_cleaned = new_data.dropna().copy()

# Create the binary target variable
new_data_cleaned['Target'] = (new_data_cleaned['Close'].shift(-1) > new_data_cleaned['Close']).astype(int)

# Drop the last row as it contains a NaN value in the Target column
eval_data = new_data_cleaned.dropna()

# Prepare features and target values
X_eval = eval_data[features]
y_eval = eval_data['Target']

# Predict using our previously trained rf_model
y_eval_pred = rf_model.predict(X_eval)

# Evaluate and print the performance metrics
print("Accuracy on 2023-2024 Data:", accuracy_score(y_eval, y_eval_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_eval, y_eval_pred))
print("\nClassification Report:\n", classification_report(y_eval, y_eval_pred))
```

# Step 5: Strategy Discussion & Critical Thinking

## 1. Model Performance Evaluation
The Random Forest Classifier showed a noticeable decline in performance when evaluated on out-of-sample data (2023–2024):
* **Validation Accuracy (2020–2022):** 54.46% (slightly above random chance).
* **Out-of-Sample Accuracy (2023–2024):** 47.26% (lower than a 50% random coin toss).
* **Class 1 (Price Increase) Recall:** Only 15%, meaning the model missed 85% of all upward price movements during 2023 (only 16 out of 109 upward days detected).
* **Class 0 (Price Decrease) Recall:** 86%, showing a heavy bias toward predicting downward or flat movements.

### Key Driver: Market Regime Shift
The primary reason for this failure is a **Market Regime Shift**:
* The training window (2020–2022) included the severe 2022 tech bear market caused by rapid interest rate hikes and inflation.
* In contrast, 2023 was a massive bull market for Big Tech, where Apple rallied aggressively from ~$125 to over ~$190, driven by the AI expansion.
* Because the model learned strong downward patterns from the prior period, it became overly pessimistic, consistently predicting price drops in a strongly trending upward market.

Given these metrics, if one were forced to build trading rules around this model:

### A. Conservative Strategy (High-Confidence Longs Only)
* **Rule:** Enter a long (buy) position only when the model predicts a price increase with high probability (e.g., confidence threshold > 75%).
* **Rationale:** Since Class 1 precision is 55%, filtering for only the strongest conviction signals helps minimize false positives.

### B. Contrarian / Short Strategy (Extreme Caution)
* **Risk Warning:** Although the model exhibited high recall (86%) for Class 0, shorting Apple shares every time the model signaled a drop in 2023 would have produced severe financial losses due to the underlying upward trend.

## 3. Critical Thinking: Is This Model Truly Useful for Financial Decision-Making?
**Verdict: NO, this model should NOT be used for real-world financial decision-making in its current state.**

**Key Reasons:**
1. **Underlying Market Characteristics (Random Walk & Non-Stationarity):** Financial markets are non-stationary and dynamic. Relying strictly on basic historical price data (OHLCV) and simple moving averages provides lagging signals rather than predictive alpha.
2. **Missing Fundamental & Exogenous Drivers:** Real-world stock performance is heavily driven by quarterly earnings reports, Fed monetary policy decisions, global supply chain factors, and market sentiment—none of which are captured by price-only technical indicators.
